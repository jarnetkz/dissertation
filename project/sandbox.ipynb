```python
# Define your function space (e.g., CG1 or CG2)
no_cells = 3
L = 1
H = 1
mesh = RectangleMesh(Point(0, 0), Point(L, H), no_cells, no_cells)
V = FunctionSpace(mesh, "CG", 1)  # or 2 for CG2
V_CG2 = FunctionSpace(mesh, "CG", 2)  # or 2 for CG2

# Total size of the state vector (total DOFs)
N = V.dim()
N_CG2 = V_CG2.dim()
print(f"Vector dimension N (Total DOFs): {N}")
print(f"Vector dimension N_CG2 (Total DOFs): {N_CG2}")

# num_vertices = mesh.num_vertices()
# print(f"Number of mesh vertices: {num_vertices}")

# # Number of elements (cells)
# num_cells = mesh.num_cells()
# print(f"Number of mesh elements/cells: {num_cells}")
```
